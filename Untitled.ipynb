# Feature Engineering 

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python
def group_countries(country):
    if country in top_countries:
        return country
    else:
        return "Other"
    
```

```python
df['Grouped_country']=df['Country'].apply(group_countries)
```

```python
df['Grouped_country'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
skewed_col = ['Study_Hours']

Other_numeric_cols= ['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks',
       'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']

ordinal_column = ['Stress_Level']

normal_columns = ['Gender', 'Country', 'Academic_Level', 'Most_Used_Platform',
       'Purpose_Of_Use','Grouped_country']

Feature_col = skewed_col+Other_numeric_cols+odinal_column+normal_columns
X=df[Feature_col]
y=df['Mental_Health_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
df['Stress_Level']
```

# Build Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer , StandardScaler , OrdinalEncoder , OneHotEncoder
from sklearn.compose import ColumnTransformer
# Skewed features
skew_pipeline = Pipeline(steps=[('log_transform',FunctionTransformer(np.log1p)),
                               ('scale',StandardScaler())])

# Numeric fetures
plain_numeric_pipeline =Pipeline(steps=[
    ('scale',StandardScaler())
])

# odinal feature 
ordinal_pipeline = Pipeline(steps=[
    ('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

# normal feture 

nominal_pipeline = Pipeline(steps=[
    ('encode' , OneHotEncoder(handle_unknown='ignore'))
])

# (konsi pipeline , konsa feature )

Preprocessor=ColumnTransformer(transformers=[
    ('Skewed_Pipeline',skew_pipeline,skewed_col),
    ('Plain_Numeric',plain_numeric_pipeline,Other_numeric_cols),
    ('Ordinal_Pipeliine',ordinal_pipeline,ordinal_column),
    ('Nominal_Pipeline',nominal_pipeline,normal_columns)
])
```

# base line model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
lr_pipeline = Pipeline(steps=[
    ('preprocessor',Preprocessor),
    ('regressor',LinearRegression())
])

lr_pipeline.fit(X_train,y_train)
lr_preds=lr_pipeline.predict(X_test)
lr_preds_train = lr_pipeline.predict(X_train)

lr_r2_testing = r2_score(y_test,lr_preds)
lr_r2_training = r2_score(y_train,lr_preds_train)
lr_ma = mean_absolute_error(y_test,lr_preds)

print(f"Accuracy of Training {lr_r2_training}")
print(f"Accuracy of Testing {lr_r2_testing}")
print(f"mean_absolute_error {lr_ma}")
```

# RandomForestRegressor

```python
 from sklearn.ensemble import RandomForestRegressor
```

```python
rand_pipeline = Pipeline(steps=[
    ('preprocessor', Preprocessor),
    ('random_forest', RandomForestRegressor(random_state=42))
])

rand_pipeline.fit(X_train,y_train)
rand_preds=rand_pipeline.predict(X_test)
rand_preds_train = rand_pipeline.predict(X_train)

rand_r2_testing = r2_score(y_test,rand_preds)
rand_r2_training = r2_score(y_train,rand_preds_train)
rand_ma = mean_absolute_error(y_test,rand_preds)

print(f"Accuracy of Training {rand_r2_training}")
print(f"Accuracy of Testing {rand_r2_testing}")
print(f"mean_absolute_error {rand_ma}")
```

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {
    "random_forest__n_estimators": [100, 200, 300],
    "random_forest__max_depth": [5, 10, 15],
    "random_forest__min_samples_split": [2, 5, 10],
    "random_forest__min_samples_leaf": [1, 2, 4]
}

random_search = RandomizedSearchCV(
    estimator=rand_pipeline,
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, y_train) 
```

```python
random_search.best_params_
```

```python
rand_pipeline = random_search.best_estimator_
rand_best_preds = rand_pipeline.predict(X_test)

print(r2_score(y_test,rand_best_preds))
print(mean_absolute_error(y_test,rand_best_preds))
```
